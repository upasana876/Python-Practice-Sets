## *1️⃣ Largest of Two Numbers*
Write a program that takes two numbers as input and prints the larger number.

```python
n1=float(input("Enter a number:"))
n2=float(input("Enter another number:"))
if n1>n2:
    print(f"{n1} is Larger than {n2}")
else:
    print(f"{n2} is Larger than {n1}")
```

## *2️⃣ Age Group Classifier*
Take a person's age as input and classify them as:
- Below 13 → Child
- 13 to 19 → Teenager
- 20 to 59 → Adult
- 60 and above → Senior Citizen

```python
Age = int(input("Enter Your Age:"))
if Age<13:
    print("The person is a Child")
elif Age<=19:
    print("The person is a Teenager")
elif Age<=59:
    print("The person is an Adult")
else:
    print("The person is a Senior Citizen")
```

## *3️⃣ Grading System*
Take marks as input and print the grade:
- 90–100 → A
- 75–89 → B
- 50–74 → C
- 35–49 → D
- Below 35 → Fail

```python
Mark = float(input("Enter Mark:"))
if Mark>=90 and Mark<=100:
    print("Grade A")
elif Mark>=75 and Mark<=89:
    print("Grade B")
elif Mark>=50 and Mark<=74:
    print("Grade C")
elif Mark>=35 and Mark<=49:
    print("Grade D")
elif Mark>=0 and Mark<35:
    print("Fail")
else:
    print("Invavlid Input")
```

## *4️⃣ FizzBuzz*
Take an integer as input:
- Divisible by 3 → Print "Fizz"
- Divisible by 5 → Print "Buzz"
- Divisible by both 3 and 5 → Print "FizzBuzz"
- Otherwise → Print the number itself

```python
num = int(input("Enter a num:"))
if num%3==0 and num%5==0:
    print("FizzBuzz")
elif num%3==0:
    print("Fizz")
elif num%5==0:
    print("Buzz")
else:
    print(f"{num}")
```

## *5️⃣ Vowel or Consonant*
Take a single character as input and determine whether it is:
- A Vowel
- A Consonant
- Not an alphabetical character

```python
c=str(input("Enter a Character:"))
v="A,E,I,O,U,a,e,i,o,u"
if c in v:
    print(f"{c} is a Vowel")
elif c not in v and c.isalpha():
    print(f"{c} is a Consonant")
else:
    print(f"{c} is not an Alphabetical Character")
```

## *6️⃣ Simple Calculator*
Take:
- Two numbers
- One operator (+, -, *, /)
Perform the operation using if-elif-else and display the result.
⚠️ Handle division by zero properly.

```python
num1=float(input("Enter a num:"))
num2=float(input("Enter a num:"))
operator=input("Enter an Operator(+,-,*,/):")
if operator=="+":
    res=num1+num2
    print(f"{num1}+{num2}={res}")
elif operator=="-":
    res=num1-num2
    print(f"{num1}-{num2}={res}")
elif operator=="*":
    res=num1*num2
    print(f"{num1}*{num2}={res}")
elif operator=="/":
    if num2==0:
        print("Error")
    else:
        res=num1/num2
        print(f"{num1}/{num2}={res}")
else:
    print("Invalid Operator")
```

## *7️⃣ Triangle Classifier*
Take three side lengths as input.
First check whether they can form a valid triangle.
If valid, classify it as:
- Equilateral Triangle
- Isosceles Triangle
- Scalene Triangle

```python
S1=float(input("Enter Side 1:"))
S2=float(input("Enter Side 2:"))
S3=float(input("Enter Side 3:"))
if S1<=0 or S2<=0 or S3<=0:
    print("Invalid Triangle")
if S1+S2>S3 and S1+S3>S2 and S2+S3>S1:
    if S1==S2==S3:
        print("Equilateral Triangle")
    elif S1==S2 and S2==S3 and S3==S1:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")     
```
