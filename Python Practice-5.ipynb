### *Question 1: Bank Account 🏦* 

Create a class "BankAccount"

* Create attributes: "account_holder", "account_number", "balance"
* Create 3 bank account objects.
* Print the account holder and balance of the second bank account.

```python
class BankAccount:
    def __init__(self,account_holder,account_number,balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance
a = BankAccount("Ashna",121,20000)
b = BankAccount("Priyanshi",125,258136)
c = BankAccount("Richa",128,1236)
print(b.account_holder, "&", b.balance)
```
