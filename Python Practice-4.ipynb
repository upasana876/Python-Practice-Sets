### Question 14
**Topic:** Exception Handling &nbsp;|&nbsp; **Type:** Code Writing

**Problem Statement:**

Write a function `safe_divide(a, b)` that uses a `try-except` block to handle division.
- If `b` is `0`, catch the `ZeroDivisionError` and print: `'Error: Cannot divide by zero'`
- Otherwise, return and print the result.
Test with `(10, 2)` and `(10, 0)`.

**Example:**  
`Input: Input: (10,2) and (10,0)`  
`Output: Output: 5.0 / Error: Cannot divide by zero`


````python
def safe_divide(a, b):
    # Write try-except here
    try:
        a=int(input("Enter a num:"))
        b=int(input("Enter another num:"))
        res=a/b
    except ZeroDivisionError:
        print('Error: Cannot divide by zero')
    except ValueError:
        print("Invalid Input")
    else:
        print("Result:",res)
    finally:
        print("Execution complete")

# Test cases
safe_divide(10, 2)
safe_divide(10, 0)
````

Use `reduce()` from the `functools` module to:
1. Find the product of all numbers in `[1, 2, 3, 4, 5, 6]`

Print both results.

**Example:**

`Output: Output: 720 / 90`


````python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]
vals = [34, 12, 78, 56, 90, 23]

# 1. Product using reduce
product = reduce(lambda a,b:a*b,numbers)
print('Product:', product)

# 2. Max using reduce
max_val = reduce(lambda a,b:a if a > b else b, vals)
print('Max:', max_val)
````

**Problem Statement:**

Chain `map()` and `filter()` together in a single expression:
- Start with `prices = [120, 450, 85, 3200, 750, 95, 1800, 60]`
- First: apply 10% discount to all prices using `map()`
- Then: filter to keep only prices **below 500** after discount
- Print the final list.

**Example:**  
`Input: Input: prices list`  
`Output: Output: prices after discount that are below 500`


````python
prices = [120, 450, 85, 3200, 750, 95, 1800, 60]

# Step 1: Apply 10% discount using map()
discounted = map(lambda x:x+x*0.1 , prices)

# Step 2: Filter prices below 500
affordable = filter(lambda x:x<500, discounted)

print(list(affordable))
````

````python

def binary_search(arr, target, low, high):
    # Base case: not found
    if low > high:
        return -1

    # Find mid
    mid = (low + high) // 2

    # Compare and recurse
    if arr[mid] == target:
        return mid

    elif target < arr[mid]:
        return binary_search(arr, target, low, mid - 1)

    else:
        return binary_search(arr, target, mid + 1, high)


arr = [2, 5, 8, 12, 16, 23, 38, 45, 72, 91]

for target in [23, 50, 2, 91]:
    result = binary_search(arr, target, 0, len(arr) - 1)
    print(f'Search {target}: index = {result}')
````

Write a **custom exception class** `InsufficientFundsError` that inherits from `Exception`.

Write a `BankAccount` class with `deposit()` and `withdraw()` methods that raise this custom exception.

**Example:**  
`Input: Input: balance=5000, withdraw 8000`  
`Output: Output: Custom exception message`


````python

# Define custom exception
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        # Set message
        self.message = f'Cannot withdraw ₹{amount}. Balance is only ₹{balance}.'

        super().__init__(self.message)


# Define BankAccount class
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        # Add to balance
        self.balance += amount

    def withdraw(self, amount):
        # Check sufficient balance
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)

        # Deduct amount
        self.balance -= amount
        print('Withdrawal successful!')
        print('Remaining balance:', self.balance)


# Test it
account = BankAccount(5000)

try:
    account.withdraw(8000)
except InsufficientFundsError as e:
    print(e)
````

````python

def describe(data):
    sorted_data = sorted(data)
    n = len(data)

    if n == 0:
        raise ValueError("Data cannot be empty")

    # Count
    count = n

    # Mean
    mean = sum(data) / n

    # Population standard deviation
    variance = sum((x - mean) ** 2 for x in data) / n
    std = variance ** 0.5

    # Min and max
    minimum = sorted_data[0]
    maximum = sorted_data[-1]

    # Median helper using sorted index positions
    def median(values):
        m = len(values)
        mid = m // 2

        if m % 2 == 0:
            return (values[mid - 1] + values[mid]) / 2
        else:
            return values[mid]

    # Quartiles: split data into lower and upper halves
    mid = n // 2

    if n % 2 == 0:
        lower_half = sorted_data[:mid]
        upper_half = sorted_data[mid:]
    else:
        lower_half = sorted_data[:mid]
        upper_half = sorted_data[mid + 1:]

    q1 = median(lower_half)
    q2 = median(sorted_data)
    q3 = median(upper_half)

    return {
        'count': count,
        'mean': round(mean, 2),
        'std': round(std, 2),
        'min': round(minimum, 2),
        '25%': round(q1, 2),
        '50%': round(q2, 2),
        '75%': round(q3, 2),
        'max': round(maximum, 2)
    }


data = [23, 45, 12, 67, 34, 89, 56, 78, 43, 91, 28, 63]

result = describe(data)

for k, v in result.items():
    print(f'{k}: {v}')
````

````python

from datetime import datetime
from functools import wraps

def audit(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')

        # Format arguments
        arguments = ', '.join(
            [repr(arg) for arg in args] +
            [f'{key}={value!r}' for key, value in kwargs.items()]
        )

        try:
            result = func(*args, **kwargs)

            log = (
                f'[{timestamp}] CALL: {func.__name__}({arguments}) '
                f'→ RESULT: {result}\n'
            )

            # Write to audit.log
            with open('audit.log', 'a', encoding='utf-8') as f:
                f.write(log)

            return result

        except Exception as e:
            log = (
                f'[{timestamp}] CALL: {func.__name__}({arguments}) '
                f'→ ERROR: {type(e).__name__}: {e}\n'
            )

            # Write error to audit.log
            with open('audit.log', 'a', encoding='utf-8') as f:
                f.write(log)

            raise

    return wrapper


@audit
def divide(a, b):
    return a / b


@audit
def square_root(n):
    if n < 0:
        raise ValueError('Cannot take square root of negative number')
    return n ** 0.5


# Test successful calls
print(divide(10, 2))
print(square_root(25))

# Test error-causing calls
try:
    divide(5, 0)
except ZeroDivisionError:
    pass

try:
    square_root(-4)
except ValueError:
    pass

# Read and print audit log
with open('audit.log', 'r', encoding='utf-8') as f:
    print(f.read())

````

````python

from functools import wraps

def retry(max_attempts=3, exceptions=(Exception,)):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    print(f'Attempt {attempt}...')
                    return func(*args, **kwargs)

                except exceptions as e:
                    print(f'Attempt {attempt} failed: {e}')

                    if attempt == max_attempts:
                        print('All attempts exhausted.')
                        raise

        return wrapper
    return decorator


# Simulate a flaky function (fails first 2 times)
call_count = 0

@retry(max_attempts=3, exceptions=(ValueError,))
def flaky_function():
    global call_count
    call_count += 1

    if call_count < 3:
        raise ValueError(f'Simulated failure #{call_count}')

    return 'Success!'


result = flaky_function()
print('Result:', result)

````

Read `'article.txt'` (use the one created in Q56 or create a new one).
Compute and display using functional programming where possible:
1. Total characters
2. Total words
3. Total lines
4. Average word length (using map and reduce)
5. Top 5 most frequent words (excluding words ≤ 3 chars)
6. Write all stats to `'stats_report.txt'`

**Example:**  
`Input: —`  
`Output: Full text statistics dashboard`


````python

from functools import reduce

# Create article.txt
article = '''Python is a powerful programming language used widely in data science.
Data science with Python has transformed how organizations analyze information.
Machine learning algorithms implemented in Python enable predictive analytics.
File handling in Python allows developers to process large datasets efficiently.
Functional programming with map filter and reduce simplifies data transformations.'''

with open('article.txt', 'w') as f:
    f.write(article)

# Read file
with open('article.txt', 'r') as f:
    content = f.read()

lines = content.splitlines()
words = content.lower().split()

# Remove punctuation
words = list(map(
    lambda word: word.strip('.,!?;:'),
    words
))
words = list(filter(lambda word: word != '', words))

# 1. Total characters
total_characters = len(content)

# 2. Total words
total_words = len(words)

# 3. Total lines
total_lines = len(lines)

# 4. Average word length using map and reduce
word_lengths = list(map(len, words))
total_length = reduce(lambda a, b: a + b, word_lengths, 0)
avg_word_length = round(total_length / total_words, 2) if total_words else 0

# 5. Top 5 most frequent words (length > 3)
long_words = list(filter(lambda word: len(word) > 3, words))

frequency = {}
for word in long_words:
    frequency[word] = frequency.get(word, 0) + 1

top_5 = sorted(
    frequency.items(),
    key=lambda item: item[1],
    reverse=True
)[:5]

# 6. Write stats to file
stats = {
    'Total Characters': total_characters,
    'Total Words': total_words,
    'Total Lines': total_lines,
    'Avg Word Length': avg_word_length,
}

with open('stats_report.txt', 'w') as f:
    f.write('=== Text Statistics Dashboard ===\n')

    for k, v in stats.items():
        f.write(f'{k}: {v}\n')

    f.write('\nTop 5 Words:\n')
    for word, count in top_5:
        f.write(f'{word}: {count}\n')

# Read and print report
with open('stats_report.txt', 'r') as f:
    print(f.read())

````

### Question 70
**Topic:** All Topics &nbsp;|&nbsp; **Type:** Interview — Capstone

**Problem Statement:**

**Capstone Challenge — Mini ETL Pipeline:**
Build a complete Extract-Transform-Load pipeline:

**Extract:** Read raw employee data from a string, save it to `'raw_employees.txt'`

**Transform:** Using functions and functional programming:
- Clean the data (handle missing/invalid values with exception handling)
- Convert salaries to float, ages to int
- Add a `tax` field: 30% of salary if salary > 50000, else 20%
- Add a `category` field: 'Senior' if age > 35, else 'Junior'

**Load:** Write the transformed data to `'processed_employees.txt'`

**Report:** Read and print a formatted summary.

**Example:**  
`Input: —`  
`Output: Full ETL pipeline with extracted, transformed, loaded data`


````python

# Raw employee data
raw = '''name,age,salary
Priya Sharma,28,55000
Rahul Verma,42,N/A
Sneha Iyer,35,72000
Arjun Nair,N/A,48000
Meena Pillai,31,91000
Kiran Kumar,38,invalid
Deep Sen,26,38000'''


# ── EXTRACT ──────────────────────────────────────────
def extract(raw_data, filename):
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(raw_data)

    print(f'Extracted to {filename}')


# ── TRANSFORM ────────────────────────────────────────
def transform(filename):
    employees = []
    errors = []

    with open(filename, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    headers = lines[0].strip().split(',')

    for line in lines[1:]:
        parts = line.strip().split(',')

        try:
            # Validate and convert data
            if len(parts) != len(headers):
                raise ValueError('Incorrect number of fields')

            name = parts[0].strip()
            if not name:
                raise ValueError('Missing name')

            age = int(parts[1])
            salary = float(parts[2])

            if age <= 0:
                raise ValueError('Age must be positive')

            if salary < 0:
                raise ValueError('Salary cannot be negative')

            # Calculate tax and category
            tax = salary * 0.30 if salary > 50000 else salary * 0.20
            category = 'Senior' if age > 35 else 'Junior'

            employees.append({
                'name': name,
                'age': age,
                'salary': salary,
                'tax': round(tax, 2),
                'category': category
            })

        except (ValueError, IndexError) as e:
            errors.append({
                'record': line.strip(),
                'error': str(e)
            })

    print(f'Transformed: {len(employees)} valid, {len(errors)} errors')
    return employees, errors


# ── LOAD ─────────────────────────────────────────────
def load(employees, filename):
    with open(filename, 'w', encoding='utf-8') as f:
        f.write('name,age,salary,tax,category\n')

        for emp in employees:
            f.write(
                f"{emp['name']},{emp['age']},"
                f"{emp['salary']:.2f},{emp['tax']:.2f},"
                f"{emp['category']}\n"
            )

    print(f'Loaded to {filename}')


# ── REPORT ───────────────────────────────────────────
def report(filename):
    with open(filename, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    print('\n========== EMPLOYEE REPORT ==========')
    print(f"{'Name':<16} {'Age':>4} {'Salary':>10} {'Tax':>10} {'Category':<8}")
    print('-' * 55)

    total_salary = 0
    total_tax = 0
    count = 0

    for line in lines[1:]:
        parts = line.strip().split(',')

        name = parts[0]
        age = int(parts[1])
        salary = float(parts[2])
        tax = float(parts[3])
        category = parts[4]

        print(
            f'{name:<16} {age:>4} {salary:>10.2f} '
            f'{tax:>10.2f} {category:<8}'
        )

        total_salary += salary
        total_tax += tax
        count += 1

    print('-' * 55)
    print(f'Valid employees: {count}')
    print(f'Total salary:    {total_salary:.2f}')
    print(f'Total tax:       {total_tax:.2f}')


# ── RUN PIPELINE ─────────────────────────────────────
extract(raw, 'raw_employees.txt')

employees, errors = transform('raw_employees.txt')

load(employees, 'processed_employees.txt')

report('processed_employees.txt')

if errors:
    print('\n--- Skipped Records ---')
    for err in errors:
        print(f"  {err['record']} → {err['error']}")
````
