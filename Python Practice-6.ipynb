## Section 1: Encapsulation Practice

Encapsulation means restricting direct access to some variables inside a class.
Use double underscore (`__`) to make a variable private.
Use getter methods to access private variables from outside the class.


### Question 1

**Problem Statement:**
Create a class `Student` with a private variable `__name`.
Write a getter method `get_name()` that returns the student name.
Create an object and call the getter method to print the name.

**Hint:** Use `__name` to declare a private variable inside `__init__`.


```python
# Write your code here
class Student:
    def __init__(self,name):
        self.__name = name
    def get_name(self):
        return self.__name
a = Student("Rahul")
print(a.get_name())
```

## Section 2: Inheritance Practice

Inheritance allows a child class to use the attributes and methods of a parent class.
Use `class Child(Parent):` to create a child class.
Use `super().__init__()` to call the parent class constructor.


### Question 9

**Problem Statement:**
Create a parent class `Animal` with an attribute `name` and a method `eat()` that prints `'{name} is eating'`.
Create a child class `Dog` that inherits from `Animal`.
Add a method `bark()` in `Dog` that prints `'{name} is barking'`.
Create a `Dog` object, call both `eat()` and `bark()`.

**Hint:** The child class inherits all methods from the parent class automatically.


```python
# Write your code here
class Animal:
    def __init__(self,name):
        self.n = name
    def eat(self):
        print(f"{self.n} is eating")
class Dog(Animal):
    def bark(self):
        print(f"{self.n} is barking")
a = Dog("Tommy")
a.eat()
a.bark()
```

### Question 14

**Problem Statement:**
Create a parent class `BankAccount` with attribute `balance` set to `10000`.
Add methods `deposit(amount)` and `withdraw(amount)` that update the balance.
Add a method `get_balance()` that returns the balance.
Create a child class `SavingsAccount` that inherits from `BankAccount`.
Add an attribute `interest_rate` in `SavingsAccount`.
Add a method `add_interest()` that increases the balance by the interest rate percentage.
Create a `SavingsAccount` object, deposit `5000`, add interest, and print the final balance.

**Hint:** `add_interest()` should modify `self.balance` which is inherited from the parent.


```python
# Write your code here
class BankAccount:
    def __init__(self):
        self.b = 10000
    def deposit(self,amount):
        self.b = self.b + amount
    def withdraw(self,amount):
        self.b = self.b - amount
    def get_balance(self):
        return self.b
class SavingsAccount(BankAccount):
    def __init__(self,interest_rate):
        super().__init__()
        self.ir = interest_rate
    def add_interest(self):
        self.b = self.b + (self.b * self.ir/100)
a = SavingsAccount(9)
a.deposit(6000)
a.add_interest()
print("Final Balance:",a.get_balance())
```

## Section 3: Polymorphism Practice

Polymorphism means the same method name can work differently for different classes.
You can define a method with the same name in multiple classes.
Each class gives its own implementation of that method.


### Question 17

**Problem Statement:**
Create two classes: `Dog` and `Cat`.

`Dog.sound()` should print `'Dog says: Woof'`.
`Cat.sound()` should print `'Cat says: Meow'`.
Create one object of each class and call `sound()` on both.

**Hint:** Same method name, different output — this is polymorphism.


```python
# Write your code here
class Dog:
    def sound(self):
        print("Dog says : Woof")
class Cat:
    def sound(self):
        print("Cat says : Meow")
a = Dog()
b = Cat()
a.sound()
b.sound()
```

## Section 4: Abstraction Practice

Abstraction means hiding internal implementation details from the user.
The user calls a simple public method without knowing what happens inside.
Private methods (using `__`) are used to hide internal steps.


### Question 24

**Problem Statement:**
Create a class `Car` with a private method `__start_engine()` that prints `'Engine started'`.
Create a public method `start()` that calls `__start_engine()` internally and then prints `'Car is ready to drive'`.
Create a `Car` object and call only `start()`.

**Hint:** The user calls `start()`. They do not need to know about `__start_engine()`.


```python
# Write your code here
class Car:
    def __start_engine(self):
        print("Engine started")
    def start(self):
        self.__start_engine()
        print("Car is ready to drive")
a = Car()
a.start()
```

Create a public method `make_coffee()` that calls all three private methods in order.
Create a `CoffeeMachine` object and call only `make_coffee()`.

**Hint:** The user only sees `make_coffee()`. All internal steps are hidden.

### Question 30

**Problem Statement:**
Create a class `DeliverySystem` with:
- A private method `__assign_driver()` that prints `'Driver assigned'`
- A private method `__calculate_route()` that prints `'Route calculated'`
- A private method `__notify_customer()` that prints `'Customer notified'`
- A public method `place_order(item, address)` that calls all three private methods in order and then prints `'Order for {item} is on its way to {address}'`

Create a `DeliverySystem` object and call `place_order('Laptop', 'MG Road, Bangalore')`.

**Hint:** Think of this like Swiggy or Zomato — the customer places an order and the internal processing is hidden.


```python
# Write your code here
class DeliverySystem:
    def __assign_driver(self):
        print("Driver assigned")
    def __calculate_route(self):
        print("Route calculated")
    def __notify_customer(self):
        print("Customer notified")
    def place_order(self,item,address):
        self.__assign_driver()
        self.__calculate_route()
        self.__notify_customer()
        self.i = item
        self.a = address
        print(f"Order for {self.i} is on its way to {self.a}")
a = DeliverySystem()
a.place_order("Laptop","MG Road, Bangalore")
```
